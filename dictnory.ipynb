```python
data = {"banana": 3, "apple": 5, "cherry": 1}

# Sort by values descending
descending_dict = dict(sorted(data.items(), key=lambda item: item[1], reverse=True))

print(descending_dict)
# Output: {'apple': 5, 'banana': 3, 'cherry': 1}
```

```python
data = {"banana": 3, "apple": 5, "cherry": 1}

# Sort by keys alphabetically
sorted_dict = dict(sorted(data.items()))

print(sorted_dict)
# Output: {'apple': 5, 'banana': 3, 'cherry': 1}
```

```python
data = {"banana": 3, "apple": 5, "cherry": 1}

# Sort by values (ascending)
sorted_by_value = dict(sorted(data.items(), key=lambda item: item[1]))

print(sorted_by_value)
# Output: {'cherry': 1, 'banana': 3, 'apple': 5}
```
