```python
# parameterize funs
# passing the values

# made a prime num fun

def isnumprime(num):
    if num <= 1:
        print("not a prime number:", num)
        return
    for i in range(2, num):
        if num % i == 0:
            print("not a prime number:", num)
            return
    print("prime number:", num)


isnumprime(6)
```

```python

def feb_Ser(n):
    series=[]
    a,b=0,1
    while a<n:
        series.append(a)
        a,b=b,a+b
    return series



print(feb_Ser(100))







```
